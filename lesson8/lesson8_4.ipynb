```python
#範本LIST

my_list = [1,2,3,4,5]
print(type(my_list))
my_list.append(6)
print(my_list)
```

```python
#請建立一個範本list
my_list = [1, 2, 3, 4, 5]
#並且建立常用的實體方法
my_list.append(6)
print(my_list)
my_list.remove(2)
print(my_list)
my_list.insert(1, 10)
print(my_list)
my_list.sort()
print(my_list)
my_list.reverse()
print(my_list)
my_list.clear()
print(my_list)
```

```python
#在list中使用for, 將每個元素*2      
my_list = [1, 2, 3, 4, 5]

for i in my_list:
    print(i * 2)    

for word in "Apple":
    print(word)

my_list = ["A","p","p","l","e"]
for word in my_list:
    print(word)
```
