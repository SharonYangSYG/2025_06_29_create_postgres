```python
#範本LIST

my_list = [1,2,3,4,5]
print(type(my_list))
my_list.append(6)
print(my_list)
```

```python
#請建立一個範本list
my_list = [1, 2, 3, 4, 5]
#並且建立常用的實體方法
my_list.append(6)
print(my_list)
my_list.remove(2)
print(my_list)
my_list.insert(1, 10)
print(my_list)
my_list.sort()
print(my_list)
my_list.reverse()
print(my_list)
my_list.clear()
print(my_list)
```

```python
#在list中使用for, 將每個元素*2      
my_list = [1, 2, 3, 4, 5]

for i in my_list:
    print(i * 2)    

for word in "Apple":
    print(word)

my_list = ["A","p","p","l","e"]
for word in my_list:
    print(word)
```

```python
#請建立一個國家code與國名字典
country_dict = {
    "US": "United States",
    "JP": "Japan",
    "CN": "China",
    "KR": "South Korea",
    "TW": "Taiwan"
}
for code in country_dict.keys():
    print(code, country_dict[code])
```

```python
for country in country_dict.values():
    print(country)
    
```

```python
for item in country_dict.items():
    print(item)
```

```python
for key, value in country_dict.items():
    print(key, value)
    
```
